# Illustrate the use of simple imputer

```python
# import libraires
from sklearn.impute import SimpleImputer
import numpy as np
```

```python
# import object using the mean strategy and missing values
imputer = SimpleImputer(missing_values=np.nan, strategy= 'mean')
data = [[12, np.nan, 35], [10, 32, np.nan,], [np.nan, 20,30]]
print("Original Data: \n", data)
```

```python
# fitting the data to the imputer object
imputer = imputer.fit(data)
```

```python
# imputing the data
data = imputer.transform(data)
print("imputed data: \n", data)
```

```python
import pandas as pd
data = {'a':[0, np.nan, 2, np.nan], 'b':[np.nan, 2, 5, 4], 'c':[-2, np.nan, np.nan, -3], 'd':[2, 1, 9, 16]}

df=pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
imputer = SimpleImputer(strategy='mean')
df_imputed=pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
print("\nDataFrame after imputing NaN values:")
print(df_imputed)
```

# Handle Missing Data with Simple Imputer

```python
#HW
import numpy as np
# Importing the SimpleImputer class
from sklearn.impute import SimpleImputer
# Imputer object using the mean strategy and 
# missing_values type for imputation
imputer = SimpleImputer(missing_values = np.nan, strategy ='mean')
data = [[12, np.nan, 34], [10, 32, np.nan], [np.nan, 11, 20]]

print("Original Data : \n", data)
# Fitting the data to the imputer object
imputer = imputer.fit(data)
# Imputing the data     
data = imputer.transform(data)
print("Imputed Data : \n", data)
```
